# 01 — corpus overview
Freeze_v1 statistics: documents by type, company, period coverage, latest-version split.

```python
# Parameters
#  freeze: data/corpus/freeze_v1.json
```

```python
import json
from collections import Counter

fr = json.load(open('data/corpus/freeze_v1.json'))
docs = fr['documents']
print('docs:', len(docs))
print('by doc_type:', dict(Counter(d['doc_type'] for d in docs)))
print('companies:', len({d['company_id'] for d in docs}))
print('unique company-periods:', len({(d['company_id'], d['fiscal_year']) for d in docs}))
print('manifest sha256:', fr['manifest_sha256'])
```
