# 00 — llm_smoke
One endpoint call: lists models and makes a minimal request when the env is configured.
No outputs are committed; re-run from a clean kernel.

```python
# Parameters
#  model: from env LLM_MODEL (default Qwen/Qwen3-8B)
#  max_tokens: 16
#  temperature: 0.0
```

```python
import os, sys
from pathlib import Path
try:
    from dotenv import load_dotenv
    load_dotenv()
except ImportError:
    pass

_root = Path.cwd().resolve()
if not (_root / 'src').is_dir():
    _root = _root.parent
sys.path.insert(0, str(_root))

_ca = os.environ.get('LLM_CA_CERT')
if _ca and not os.path.isabs(_ca) and not os.path.exists(_ca):
    _cand = _root / _ca
    if _cand.exists():
        os.environ['LLM_CA_CERT'] = str(_cand)

if not all(os.environ.get(k) for k in ('LLM_BASE_URL', 'LLM_API_KEY', 'LLM_MODEL', 'LLM_CA_CERT')):
    print('env not configured; skipping. Create .env with LLM_BASE_URL, LLM_API_KEY, LLM_CA_CERT, LLM_MODEL.')
else:
    from src.llm import client
    print('models:', client.list_models())
    resp = client.chat([{'role': 'user', 'content': 'Reply with the single word: ok'}],
                       max_tokens=16, temperature=0.0, cache=False)
    print('status:', resp.status, 'cached:', resp.cached, 'prompt_version:', resp.prompt_version)
    print('text:', resp.text)
```
